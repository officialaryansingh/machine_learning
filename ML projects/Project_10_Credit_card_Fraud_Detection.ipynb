```python
# impoting the important dependencies
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
# loading the dataset to pandas DataFrame
credit_card_data = pd.read_csv('/content/credit_data.csv')
```

This dataset is Highly unbalanced

0 --> Normal Transaction

1--> Fraud Transaction

```python
# seperating the data for analysis
legit = credit_card_data[credit_card_data.Class == 0]
fraud = credit_card_data[credit_card_data.Class == 1]
```

```python
print(legit.shape)
print(fraud.shape)
```

```python
# statistical information
legit.Amount.describe()
```

```python
fraud.Amount.describe()
```

```python
# compare the values for both transaction
credit_card_data.groupby('Class').mean()
```

Under Sampling

Build a sample dataset containing similar distribution of normal trancation and
fraudulent transaction

number of Fraudlent Transactions --> 470

```python
legit_sample = legit.sample(n=470)
```

```python
# concatenating two DataFrame
new_dataset = pd.concat([legit_sample,fraud],axis = 0)
```

```python
new_dataset.head()
```

```python
new_dataset['Class'].value_counts()
```

```python
new_dataset.groupby('Class').mean()
```

Splitting the data into Features & Targets

```python
X = new_dataset.drop(columns = 'Class',axis = 1)
Y = new_dataset['Class']
```

```python
print(X)
```

```python
print(Y)
```

Splitting the data in Training data and Testing Data

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size = 0.2,stratify = Y,random_state = 2)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

Model Training

Logistic Regression

```python
model = LogisticRegression()
```

```python
model.fit(X_train,Y_train)
```

Model Evalution

Accuracy Score

```python
# accuracy on trainig data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction,Y_train)
```

```python
print('Accuracy on training on Training data : ', training_data_accuracy)
```

```python
# accuracy for testing data
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction,Y_test)
```

```python
print('Accuracy score on test data : ',test_data_accuracy)
```

Building a predictive system

```python
user_input = input("Enter the here (comma seperated) : ")

# converting the input into the list of floats (or ints)
input_data = [float(x) for x in user_input.split(',')]

# change the data into the numpy array
input_data_as_numpy_array = np.asarray(input_data)

# reshaping the numpy array as  we are only prediticting on instance
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)

prediction = model.predict(input_data_reshaped)
print(prediction)

if(prediction == 0):
  print("The transaction is legit")

else:
  print("The transaction is fraud")
```
