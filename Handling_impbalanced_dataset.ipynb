**Imbalanced Dataset**

A  dataset with an unequal class distribution

```python
# importing the dependencies
import numpy as np
import pandas as pd
```

```python
# laoding datasest to pandas dataFrame
credit_card_data = pd.read_csv('/content/credit_data.csv')
```

```python
# First 5 rows of the dataFrame
print(credit_card_data.head())
```

```python
#distribution of the two classes
credit_card_data['Class'].value_counts()
```

This is Highly Imbalaced Dataset

0--> Legit Transactions

1--> Fraudulent Transactions

```python
# seperating the legit and fraudulent transactions
legit = credit_card_data[credit_card_data.Class == 0]
fraud = credit_card_data[credit_card_data.Class == 1]
```

```python
print(legit.shape)
print(fraud.shape)
```

**Under-saampling**

Build a sample dataset containing similar distribution of legit & fraudulent Transactions

Number of fraudulent transactions --> 492

```python
legit_sample = legit.sample(n=492)
```

```python
print(legit_sample.shape)
```

Concatenate the Two DataFrame

```python
new_dataset = pd.concat([legit_sample,fraud],axis = 0)
```
